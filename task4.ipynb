```python
!pip install nltk textblob
import pandas as pd
import nltk
from textblob import TextBlob
import matplotlib.pyplot as plt
import seaborn as sns
print("Libraries imported successfully")
data = {
    "review": [
        "I love this product, it is amazing!",
        "This is the worst experience I ever had.",
        "The service was okay, not bad.",
        "Absolutely fantastic and wonderful!",
        "I am very disappointed and unhappy.",
        "It is average, nothing special."
    ]
}
df = pd.DataFrame(data)
print("\nSample Reviews:")
print(df)
def get_sentiment(text):
    analysis = TextBlob(text)
    polarity = analysis.sentiment.polarity
    if polarity > 0:
        return "Positive"
    elif polarity < 0:
        return "Negative"
    else:
        return "Neutral"
df["sentiment"] = df["review"].apply(get_sentiment)
print("\nReviews with Sentiment:")
print(df)
sns.countplot(x="sentiment", data=df)
plt.title("Sentiment Analysis Result")
plt.xlabel("Sentiment")
plt.ylabel("Count")
plt.show()
print("\nTask-4 Sentiment Analysis completed successfully")
```
